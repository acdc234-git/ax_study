### 1. 딕셔너리(Dictionary) 자료구조

- Key - Value 쌍 구조
- Key : Value를 찾기 위한 키워드, 중복 x
- Value : Key로 찾은 값, 중복 O
- {"키":값,"키":값,...}

- 추가(CREATE) : 없는 키 이름으로 값을 대입하면 추가
- 조희(READ) : ['키 이름'] - 키가 없으면 KeyError가 발생
    - get("키 이름") - 키가 없으면 None 반환
    - get("키 이름", "기본값") - 키 이름으로 조회가 되지 않으면 설정된 기본값으로 대체
    - keys() : 모든 키를 담은 View, 읽기 전용
    - values() : 모든 값을 담은 View, 읽기 전용
    - items() : (키,값) 목록을 담은 View
- 수정(UPDATE) : 이미 있는 키값에 대입하면 수정
    - update({key,value,key,value,....})
- 삭제(DELETE) :
    del items['키']
    celar(): 전체 비우기
다른 키-값 쌍 자료형 -> 딕셔너리로 변환
dict(...)

```python
user = {
        "email":"test@test.org",
        "password":1234,
        "name":"김철수"
        }
```

### 3. 집합(Set) 자료구조

- 중복을 허용하지 않는다.
- 순서가 결정되지 않는다.
={값1,값2,값3,...}

#### 집합 연산
- 합집합: a|b -> a와 b의 합집합(a.union(b))
- 교집합: a&b -> a.intersection(b)
- 차집합: a-b -> a.difference(b)
- 대칭차집합: a^b -> a.symmetric_difference(b)

#### 집합 자료로 형변환
- set(리스트 또는 튜플)

```python
a = {10,10,10,20,30}
a
```

```python
a[0]
```

```python
a = {10,20,30,40}
b = {30,40,50,60}

print("합집합:",a|b)
print("교집합:",a&b)
print("차집합:",a-b)
print("대칭차집합:",a^b)
```
