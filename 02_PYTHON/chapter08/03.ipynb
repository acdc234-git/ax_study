### 1. 클래스 변수(정적 변수), 인스턴스 변수

```python
class User:
    pass

user = User()
user.name = '김철수'
user.age = 20

print(user.name, user.age)
```

```python
class User:
    count = 0
```

```python
user1 = User()
user2 = User()
print(user1.count, user2.count)
```

### 2.@classmethod, @staticmethod

- 객체를 생성하지 않아도 접근할 수 있는 메서드를 정의할때

```python
class Caluculator:
    count = 0

    #clz -> Calculater 클래스 객체 참조 변수
    #clz -> 클래스 변수, 다른 클래스 메서드를 접근하기 위한 목적
    @classmethod
    def add(clz, num1, num2):
        clz.count +=1
        print(f'호출 횟수: {clz.count}')
        return num1 + num2

    @staticmethod
    def minus(num1, num2):
        return num1 - num2
```

```python
Caluculator.add(10,20)
```
