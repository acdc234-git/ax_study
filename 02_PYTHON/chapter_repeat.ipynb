```python
import time

def download_page(data,num):
    print(f"{data} 시작")
    time.sleep(num)
    print(f"{data} 완료")

def main():
    start = time.time()
    download_page('메인 페이지 데이터',2)
    download_page('상세 페이지 데이너',2)
    end = time.time()
    result = round((end - start),1)

    print(f"총 소요 시간{result}")

main()
```

```python
import asyncio
import time

async def download_page_async(data, second):
    print(f"{data} 시작")
    await asyncio.sleep(second)
    print(f"{data} 완료")

async def main():
    start = time.time()
    await asyncio.gather(download_page_async('메인 페이지 데이터',2), download_page_async('상세 페이지 데이터',2))
    end = time.time()
    print(f"총 소요 시간: {round(end-start)}")
   
await main()
```

```python
async def fetch_weather(place, second):
    print(f"{place}의 날씨를 {second}초 동안 수집 합니다")
    await asyncio.sleep(second)
    print(f"{place}의 데이터 수집이 완료되었습니다")

async def main():
    start = time.time()
    await asyncio.gather(fetch_weather('서울',1.5),fetch_weather('뉴욕',2.5),fetch_weather('런던',1.0))
    end = time.time()
    print(f"총 소요 시간: {round(end-start,1)}")

await main()
```
