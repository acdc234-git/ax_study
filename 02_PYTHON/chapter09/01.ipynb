### 2. all() 과 any()

- all(iterable) : 모든 요소가 True로 인식될때 -> True, 예외) 빈 리스트([]) - True
- any(iterable) : 요소중 하나라도 True 인식이 되면 -> True

```python
all([True,1,' '])
```

```python
all([False,False])
```

```python
all([])
```

```python
any([False, 1, 0])
```

```python
any([])
```

### 3. len(), sum(), round()

- len() : 문자열 -> 문자열의 길이, 리스트, 튜플 -> 요소의 갯수
- sum(iterable) : 합계 (요소는 숫자로만 구성)
- round() : 반올림, Math: ceil(..): 올림, floor(..): 버림

```python
num = 10.56
print(round(num,1))

import math
print(math.ceil(10.2))
print(math.floor(num))
```

### 4. mean()과 median()

- 모듈에 포함딘 표주라이브러리
- mean(iterbale): 평균
- median(iterabel): 중앙값

```python
import statistics

nums = [10,15,20,8,7,1000000]

print("평균:", statistics.mean(nums), "중앙값: ", statistics.median(nums))
```

### 5. map()과 filter()

- map() : 변환
- filter() : 판별 -> 정제

-> 리스트 컴프리헨션으로 대체 가능

```python
nums = [*range(1,11)]

nums2 = [x**2 for x in nums]
print(nums2)

nums3 = [x for x in nums if x % 2 == 1]
print(nums3)
```

### 6. enumerate()

```python
fruits = ['사과', '오렌지', '멜론', '망고']

num = 1
for fruit in fruits:
    print(fruit)
    num += 1
```

```python
for item in enumerate(fruits, start=1): # (순서 번호, 요소)
    #print(item)
    print(item, fruits)
```

### 7. zip

참고 - itertools 모듈 zip_longest

```python
persons = ['김철수','김영희','김민수']
fruits = ['사과','오렌지','멜론','망고']
```

```python
items = []
for i, person in enumerate(persons):
    items.append((person,fruits[i]))

items
```

```python
items = zip(persons, fruits)

list(items)
```

### 8. sorted()

- 리스트.sort(): 원본 데이터를 정렬
- sorted(Iterable): 원본 데이터는 유지, 새로운 객체를 생성해서 정렬(불변객체)

```python
import random

nums = [*range(1,11)]
random.shuffle(nums)

nums
```

```python
nums.sort()

nums
```

```python
nums_sorted = sorted(nums)

print('nums_sorted: ', nums_sorted)
print('nums: ',nums)
```

### 9. min(), max()

- min() : 최소값 / 오름차순 정렬 후 가장 처음 요소
- max() : 최대값 / 오름자순 정렬 후 가장 마지막 요소
