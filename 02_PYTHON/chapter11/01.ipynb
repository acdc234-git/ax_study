## 동기(Sync) vs 비동기(Async)의 이해

```python
# 동기식 함수 (서브 루팅)
import time

def work(title, seconds):
    time.sleep(seconds)
    print(title,"완료!")

start = time.perf_counter() # 작업 시작 시간

work("A작업",2)
work("B작업",3)
work("C작업",1)

end = time.perf_counter() # 작업 종료 시간
print('걸린시간:', end - start)

```

```python
# 비동기 함수 (코루틴)
import asyncio
import time

async def async_work(title, seconds):
    asyncio.sleep(seconds)
    print(title,"완료!")

async def main():
   
    asyncio.gather(
        async_work('A작업',2),
        async_work("B작업",3),
        async_work('C작업',1)
    )
    
    print("걸린시간:",end-start)
    
start = time.perf_counter()
await main()
end = time.perf_counter()
```
