## 1. 파이썬의 기본 자료형 (Primitive Data Types)

 - 자료 : 연산에 필요한 값(데이터)
 - 자료형 : 값의 유형(데이터의 종류)

 #### 자료형
 - 숫자형 : 정수형(int, 10, 20, ...), 실수형(float, 10.123, ..)
 - 논리형 : 참(True), 거짓(False)
 - 문자형 : 문자, 문장, '', "" 감싸서 선언, """...""" - 여러줄 문장
 - 값 없음 : None

```python
money = None
```

```python
message = """안녕하세요.
반갑습니다.
"""
type(message)
```

## 2. 명시적 형변환(Explicit Type Conversion)

연산은 같은 자료형끼리 가능

형변환 함수
- int(..) : 정수로 변환
- float(..) : 실수로 변환
- bool(...) : True, False 변환
- str(...) : 문자형으로 변환

```python
num = "100"
type(num)
```

```python
int(num) + 200
```

```python
"현재 시간은 " + str(10) + "시"
```

## 3. 묵시적 형변환(Implicit Type Conversion)

```python
num1 = 100
num2 = 3.123

type(num1),type(num2)
```

```python
# num3 = num1 + num2 # num1이 int -> float 자동 형변환 (묵시적 형변환)
num3 = num1 + int(num2)

num3, type(num3)
```

## 4. 형변환 시 주의해야 할 예외와 에러(ValueError)

```python
time = "10시"
if time.isdigit():
    int(time) + 1

print(time)
```
