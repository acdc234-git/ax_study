### 1. 함수의 기본 정의와 작동 방식

#### 정의 방법
```
def 함수이름(매개변수,...)
    # 처리 부분

    return 반환값 
```
#### 함수를 호출하는 방법(실행하는 방법)
이름(...) -> 엔터

#### 목적
코드의 재활용

#### return
- 반환값 정의
- 함수의 종료

````python
def calc(x):
    y = x * 2 + 1

    return y
````

````python
result = calc(10)
result
````

````python
def add(num1, num2):
    num3 = 30

    result = num1 + num2 + num3 # 지역변수 result

    return result
````

````python
result = add(10,20) # 전역변수 result

result
````

````python
def func(num):
    result = add(10,20)
    result += num
    return result
````

````python
def add(num):
    if num == 999:
        print('종료')
        return
    result = num ** 2

    return result
````

### 2. 매개변수(Parameter)와 인수(Argument)

- 매개변수 : 함수 호출시 값을 담아주는 변수
- 인수 : 호출시 사용된 값 (10)

### 3. 위치 인수(Positional Arguments)와 키워드 인수(Keyword Arguments)

- 위치 인수 : 매개변수의 순서대로 값을 지정

````python
def add(num1,num2):

    return num1 + num2
````

````python
add(10, 20) #위치 인수
````

````python
add(num1=10,num2=20) #키워드 인수
````

````python
add(num2=20,num1=10) #키워드 이름이 중요, 순서는 상관 없음
````

````python
def add(num1, num2, name, mobile):
    return f"{num1+num2}, {name=}, {mobile=}"
````

````python
add(10,20,mobile='010-1000-1000', name = '김영희')
````

### 4. 기본값 매개변수(Default Parameter)

- 매개변수에 값을 할당하지 않아도 기본적으로 가질수 있는 값 설정
- 기본값이 있는 매개변수는 오른쪽 끝부터 차례대로 정의


````python
def add(num1, num2, name, mobile= '010'):
    return f"{num1+num2}, {name=}, {mobile=}"
````

````python
add(10,20,'김영희')
````

### 5. 가변 인수(*args)와 키워드 가변 인수(**kwargs)

- *args: 가변적인 위치 인수, 매개변수 갯수 상관없이 지정 가능, 키워드 매개변수로는 사용 불가

````python
def add(num1,num2,num3,num4):
    return num1+num2+num3+num4

add(10,20,30,40)
````

````python
def add(*nums):
    #print(nums,type(nums))
    total = 0
    for num in nums:
        total += num
    return total
````

````python
print(add(10,20))
print(add(10,20,30))
print(add(10,20,30,40))
````

````python
def print_user_info(**info):
    print(info)
````

````python
print_user_info(name='김철수',age=20)
````

````python
def print_user_info(name, age, **info):
    print("name:", name)
    print("age:", age)
    print(info)
````

````python
print_user_info('김철수',20, mobile = '010-1000-1000')
````

````python
def print_user_info(name,age,*args,**kwargs):
    print("name:", name)
    print("age:", age)
    print(args)
    print(kwargs)
````

````python
print_user_info('김철수',20, 'A',20,mobile = '010-1000-1000')
````
