```python
import os
from supabase import create_client, Client

url: str = os.getenv("SUPABASE_URL")
key: str = os.getenv("SUPABASE_PUBLISHABLE_KEY")

supabase: Client = create_client(url,key)

supabase
```

```python
def test_order_transaction():
    # 테스트 페이로드 구성 (의도적으로 두 번째 상품 ID를 비정상 UUID로 입력)
    payload = {
        "p_order_name": "홍길동",
        "p_order_email": "hong@example.com",
        "p_order_phone": "010-1234-5678",
        "p_receiver_name": "이순신",
        "p_receiver_phone": "010-9876-5432",
        "p_zipcode": "06234",
        "p_address": "서울시 강남구 테헤란로 123",
        "p_address_sub": "7층",
        "p_delivery_memo": "문 앞에 놓아주세요.",
        "p_user_id": "aaaaaaaa-aaaa-aaaa-aaaa-aaaaaaaaaaaa",
        "p_items": [
            { "product_id": "11111111-1111-1111-1111-111111111111", "quantity": 1 }, # 정상 상품 (키보드)
            { "product_id": "99999999-9999-9999-9999-999999999999", "quantity": 2 }  # 존재하지 않는 가짜 상품
        ]
    }

    try:
        # Supabase RPC 원격 함수 실행
        response = supabase.rpc("create_order_transaction", payload).execute()
        print("주문 성공 완료! 응답 데이터:", response.data)

    except Exception as e:
        # DB에서 RAISE EXCEPTION이 발생하면 파이썬 Exception으로 포착됩니다.
        print("에러:", e)
```
