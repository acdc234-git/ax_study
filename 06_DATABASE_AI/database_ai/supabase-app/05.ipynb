## 메서드 체이닝
- 선언적 방식

```
supabase.table(...)
    .select(...)
    .eq(...)
    .execute()
```

````python
class SupabaseEx:
    def __init__(self):
        self.setting =[]

    def table(self,table):
        self.setting.append(table)
        
        return self

    def select(self, fields):
        self.setting.append(fields)

        return self # 생성된 동일 객체

    def eq(self, name, value):
        self.setting.append(f"{name}={value}")

        return self

    def execute(self):
        print(self.setting)
        print("실행...")
        return "처리된 결과..."
````

````python
supabase = SupabaseEx()

response = (
    supabase.table("users")
        .select("*")
)
````

````python
## 빌더 패턴 / 생성 패턴
````

````python
class Product(object):
    # 내부 식별자
    _sentinel = object()

    # 클래스 외부에서 직접 생성은 차단, 
    def __init__(self, token, name, price, description):

        if token is not Product._sentinel: # 외부에서 직접 생성한 경우
            raise Exception("직접 생성할 수 없습니다.")
        
        self.__name = name
        self.__price = price
        self.__description = description

    def name(self):
        return self.__name

    def price(self):
        return self.__price

    def description(self):
        return self.__description

    def __repr__(self):
        return f"Product(name='{self.__name}, price='{self.__price}', description = '{self.__description}')"

    # Product 객체를 조립하고 생성하는 클래스
    class Builder:
        def name(self,name):
            self.__name = name

            return self

        def price(self, price):
            self.__price = price

            return self

        def description(self, description):
            self.__description = description

            return self

        # 객체를 생성
        def build(self):
            return Product(Product._sentinel, self.__name, self.__price, self.__description)
````

````python
product = (
        Product.Builder()
            .name("아이폰")
            .price(10000)
            .description("설명")
            .build()
)

product
````

````python
print(product.name(),product.price(), product.description())
````

````python
# 데이터 클래스 - @dataclass 

from dataclasses import dataclass

@dataclass()
class Product:
    name: str
    price: int = 0
    description: str = ""
````

````python
product = Product("아이폰",10000,"설명")

product
````

````python
product = Product(
    name="아이폰",
    description="설명",
    price=10000
)

product
````
